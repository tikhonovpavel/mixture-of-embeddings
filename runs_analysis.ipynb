```python
import os
import json
import pandas as pd
from pathlib import Path


checkpoints_dir = Path('checkpoints')
runs = []

for run_dir in sorted(checkpoints_dir.iterdir()):
    if not run_dir.is_dir():
        continue
    
    config_path = run_dir / 'training_config.json'
    
    if not config_path.exists():
        continue
    
    checkpoints = list(run_dir.rglob('*.pt'))
    if not checkpoints:
        continue
    
    with open(config_path, 'r') as f:
        config = json.load(f)
    
    config['run_name'] = run_dir.name
    runs.append(config)

df = pd.DataFrame(runs)

first_cols = ['run_name', 'model_name']
other_cols = [col for col in df.columns if col not in first_cols]
other_cols_sorted = sorted(other_cols, key=lambda col: df[col].nunique(), reverse=True)

df = df[first_cols + other_cols_sorted]

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)

df
```
